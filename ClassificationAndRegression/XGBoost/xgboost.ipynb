
# Gradient Boosting

## 1. What is Gradient Boosting?

Gradient Boosting is an ensemble machine learning technique that combines multiple weak learners (usually Decision Trees) sequentially to create a strong predictive model.

Each new tree tries to correct the errors made by the existing ensemble by following the negative gradient of a loss function.

**Key idea:** Instead of building independent trees, Gradient Boosting builds trees one after another, with each new tree improving the overall prediction.

## 2. How does Gradient Boosting work?

Consider a house price prediction problem.

| House | Actual Price (₹ Lakhs) |
|---|---:|
| A | 50 |
| B | 70 |
| C | 90 |

### Step 1: Initial Prediction (Tree 1)

Assume the initial model predicts the average price for every house.

| House | Actual | Predicted | Residual (Actual - Predicted) |
|---|---:|---:|---:|
| A | 50 | 70 | -20 |
| B | 70 | 70 | 0 |
| C | 90 | 70 | +20 |

The first model makes prediction errors.

### Step 2: Train the Second Tree

The second tree learns to predict the residuals (errors) of the first model.

| House | Residual | Tree 2 Prediction |
|---|---:|---:|
| A | -20 | -15 |
| B | 0 | 0 |
| C | +20 | +15 |

### Step 3: Combine Predictions

The new prediction is calculated by adding the correction from the second tree to the previous prediction.

Assuming the learning rate is 1:

| House | Tree 1 | Tree 2 | Updated Prediction |
|---|---:|---:|---:|
| A | 70 | -15 | 55 |
| B | 70 | 0 | 70 |
| C | 70 | +15 | 85 |

The predictions are now closer to the actual prices.

### Step 4: Repeat

The next tree learns from the remaining errors. This process continues until the model reaches a specified number of trees or another stopping condition.

**Important:** This residual-based explanation applies directly to squared-error regression. More generally, Gradient Boosting fits each new tree to the negative gradients of the selected loss function.

---

## 3. Why is it called Gradient Boosting?

The name has two parts:

- **Boosting:** Combining multiple weak learners sequentially to create a stronger model.
- **Gradient:** Using the negative gradient of a loss function to determine how the model should improve.

The gradient indicates the direction of increasing loss. Gradient Boosting uses its negative direction to reduce the loss.

## 4. Mathematical Foundation

### A. Loss Function

A loss function measures the difference between actual and predicted values.

For regression, Mean Squared Error (MSE) is commonly used:

\[
L = \frac{1}{n}\sum_{i=1}^{n}(y_i-\hat{y}_i)^2
\]

Where:

- \(y_i\) = Actual value
- \(\hat{y}_i\) = Predicted value
- \(n\) = Number of observations

The objective is to minimize the loss.

### B. Residuals and Negative Gradients

For squared-error regression, the negative gradient is proportional to the residual:

\[
r_i = y_i-\hat{y}_i
\]

This explains why each new tree learns to predict the residuals of the existing model.

For other loss functions, the negative gradient may not equal the ordinary residual.

### C. Prediction Update

The general prediction update is:

\[
F_m(x)=F_{m-1}(x)+\eta h_m(x)
\]

Where:

- \(F_m(x)\) = Updated model prediction
- \(F_{m-1}(x)\) = Previous model prediction
- \(h_m(x)\) = New tree's predicted correction
- \(\eta\) = Learning rate
- \(m\) = Current boosting iteration

The learning rate controls how much each new tree contributes to the final prediction.

---

## 5. Learning Rate

Learning rate controls the contribution of each new tree.

Suppose a new tree predicts a correction of +20.

| Learning Rate | Actual Correction Added |
|---|---:|
| 1.0 | +20 |
| 0.5 | +10 |
| 0.1 | +2 |
| 0.01 | +0.2 |

- A higher learning rate allows faster updates but may increase the risk of overfitting.
- A lower learning rate makes updates smaller and generally requires more trees.
- Learning rate and number of estimators should be tuned together.

---

## 6. Gradient Boosting vs Random Forest

| Feature | Random Forest | Gradient Boosting |
|---|---|---|
| Technique | Bagging | Boosting |
| Tree construction | Independent | Sequential |
| Main idea | Reduce variance through averaging | Improve predictions using gradients |
| Tree combination | Average / Majority voting | Additive predictions |
| Learning rate | No | Yes |
| Error correction | No sequential correction | Each stage improves the ensemble |
| Overfitting control | Averaging and tree constraints | Learning rate, tree constraints and other regularization methods |

**Remember:**
- Random Forest builds multiple independent trees and combines their predictions.
- Gradient Boosting builds trees sequentially, with each new tree improving the existing ensemble.

---

## 7. Important Hyperparameters

| Hyperparameter | Description |
|---|---|
| `n_estimators` | Number of boosting stages (trees) |
| `learning_rate` | Contribution of each new tree |
| `max_depth` | Maximum depth of each decision tree |
| `min_samples_split` | Minimum samples required to split an internal node |
| `min_samples_leaf` | Minimum samples required at a leaf node |
| `subsample` | Fraction of training samples used for each boosting stage |

### Important relationship

A smaller learning rate generally requires a larger number of estimators.

Too many trees or overly complex trees can cause overfitting.

---

## 8. Advantages

1. Can model complex nonlinear relationships.
2. Often performs very well on structured/tabular datasets.
3. Supports different loss functions.
4. Can capture interactions between features.
5. Does not require feature scaling in the same way as distance-based algorithms.

## 9. Limitations

1. Training is sequential, so it is less parallelizable across boosting stages.
2. Can overfit if trees are too complex or too many.
3. Requires hyperparameter tuning.
4. Can be sensitive to noisy data and outliers depending on the loss function.
5. Training can become computationally expensive with many boosting stages.

| LightGBM | Efficient gradient boosting framework designed for speed and scalability |

---

## 11. Key Takeaways

- Gradient Boosting is an ensemble learning technique.
- It uses multiple weak learners, usually Decision Trees.
- Trees are trained sequentially.
- Each new tree learns from the negative gradients of the loss function.
- For squared-error regression, these negative gradients are the residuals.
- Learning rate controls the contribution of each tree.
- The objective is to minimize the loss function.

# Gradient Boosting - Mathematical Formulas

### 1. Mean Squared Error (MSE)

Measures the difference between actual and predicted values.

$$
MSE = \frac{1}{n}\sum_{i=1}^{n}(y_i-\hat{y}_i)^2
$$

Where:
- $y_i$ = Actual value
- $\hat{y}_i$ = Predicted value
- $n$ = Number of observations

### 2. Residual (Prediction Error)

For squared-error regression, the residual is the difference between actual and predicted values.

$$
r_i = y_i - \hat{y}_i
$$

### 3. Negative Gradient

Gradient Boosting uses the negative gradient of the loss function to determine the correction required.

For squared-error loss:

$$
L_i = \frac{1}{2}(y_i-\hat{y}_i)^2
$$

$$
r_i = -\frac{\partial L_i}{\partial \hat{y}_i}
$$

Therefore:

$$
r_i = y_i-\hat{y}_i
$$

### 4. Prediction Update Formula

Each new tree contributes a correction to the previous prediction.

$$
F_m(x) = F_{m-1}(x) + \eta h_m(x)
$$

Where:
- $F_m(x)$ = Updated prediction
- $F_{m-1}(x)$ = Previous prediction
- $h_m(x)$ = New tree's correction
- $\eta$ = Learning rate
- $m$ = Current boosting iteration

### 5. Learning Rate

The learning rate controls how much correction is added by each new tree.

$$
New\ Prediction = Previous\ Prediction + (\eta \times Correction)
$$

Example:

If the previous prediction is 70, the new tree predicts a correction of 20, and the learning rate is 0.1:

$$
New\ Prediction = 70 + (0.1 \times 20)
$$

$$
New\ Prediction = 72
$$

### 6. Overall Objective

Gradient Boosting aims to minimize the loss function by adding trees sequentially.

$$
F^*(x) = \arg\min_F \sum_{i=1}^{n} L(y_i,F(x_i))
$$

**Remember:** Gradient Boosting minimizes prediction error by repeatedly adding new trees that learn from the negative gradients of the loss function.
