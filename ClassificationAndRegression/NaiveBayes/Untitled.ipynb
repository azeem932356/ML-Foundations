## Naive Bayes Theorem

Naive Bayes is a supervised machine learning algorithm based on **Bayes' Theorem**. It is used for classification by calculating the probability of an observation belonging to each class and predicting the class with the highest probability.

### 1. Bayes' Theorem

$$
P(A|B) = \frac{P(B|A) \times P(A)}{P(B)}
$$

Where:

- $P(A|B)$ = Posterior probability: probability of A given B.
- $P(B|A)$ = Likelihood: probability of B given A.
- $P(A)$ = Prior probability: initial probability of A.
- $P(B)$ = Evidence: probability of B.

### 2. Naive Bayes Classification Formula

For multiple features, Naive Bayes assumes that features are conditionally independent given the class.

$$
P(C|x_1,x_2,\ldots,x_n)
\propto
P(C)\prod_{i=1}^{n}P(x_i|C)
$$

Where:

- $C$ = Target class.
- $x_1,x_2,\ldots,x_n$ = Input features.
- $P(C)$ = Prior probability of the class.
- $P(x_i|C)$ = Conditional probability of a feature given the class.
- $\prod$ = Multiplication of probabilities.
- $\propto$ = Proportional to.

### 3. How Naive Bayes Works

1. Calculate the prior probability of each class.
2. Calculate the likelihood of each feature for each class.
3. Multiply the prior by the feature likelihoods.
4. Compare the resulting class scores.
5. Predict the class with the highest probability.

### 4. Why Is It Called Naive?

It assumes that input features are conditionally independent of one another, given the target class. This assumption simplifies probability calculations, although real-world features may be correlated.

### 5. Types of Naive Bayes

| Algorithm | Suitable Data |
|---|---|
| Gaussian Naive Bayes | Continuous numerical features |
| Multinomial Naive Bayes | Counts and frequencies |
| Bernoulli Naive Bayes | Binary features |
| Categorical Naive Bayes | Categorical features |

### 6. Laplace Smoothing

Laplace smoothing prevents zero probabilities when a feature has not appeared in a particular class.

$$
P(x_i|C) =
\frac{Count(x_i,C)+\alpha}
{Count(C)+\alpha V}
$$

Where:

- $\alpha$ = Smoothing parameter (commonly 1).
- $V$ = Number of possible feature values or vocabulary terms, depending on the model.
